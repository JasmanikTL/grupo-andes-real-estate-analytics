# Grupo Andes Inmobiliario — Commercial Analytics

## Project overview
This project examines real-estate sales performance in 2023–2024 through a Power BI dashboard. The analysis covers sales, revenue, commissions, property categories, sales channels, year-over-year trends, and repeat purchases.

## Data model
- `hecho_ventas_propiedades`: transaction-level sales fact table
- `dim_propiedades`: property attributes
- `dim_clientes`: customer segments
- `dim_fecha`: calendar dimension described in the Power BI workflow

## Analytical workflow
1. Inspect data types, missing values and dimension keys.
2. Model sales with property, customer and date dimensions.
3. Analyze revenue, transaction count, average sale value and commissions.
4. Compare property types, cities, channels and customer segments.
5. Explore year-over-year changes and customer cohorts.

**Implementation note:** The original notebook is primarily a written project report; dashboard logic was developed in Power BI. No executable Python analysis is claimed here.



## 📝 Paso 6: Resumen Ejecutivo

### 🔑 Hallazgos clave

- Durante el periodo **2023–2024**, se generaron **$6.01 mil millones** en ingresos mediante **8,500 ventas**.
- Las **casas** fueron el tipo de propiedad que generó mayor ingreso, con **$2.24 mil millones**, seguidas de los inmuebles comerciales con **$1.92 mil millones** y los departamentos con **$1.85 mil millones**.
- **Ciudad de México** concentró el mayor ingreso, con aproximadamente **$3.24 mil millones**, frente a **$2.77 mil millones** de Bogotá.
- El canal **Corredor** generó considerablemente más ingresos que el canal **Directo**.
- En **2024**, los ingresos alcanzaron **$3.16 mil millones**, lo que representó un crecimiento de **11.14 %** respecto al año anterior.

### 📊 Métricas principales

- **Ingreso total:** $6.01 mil millones.
- **Cantidad de ventas:** 8,500.
- **Ticket promedio:** $707.35 mil.
- **Comisión total:** $200.63 millones.
- **Crecimiento YoY 2024:** 11.14 %.

### 💡 Insights accionables

- El segmento **Primera vez** concentra el mayor volumen de operaciones, con más de **5 mil ventas**, por encima de los segmentos Inversionista y Alto patrimonio.
- Los **departamentos** presentan el mayor número de operaciones, con **5,105 ventas**, pero tienen el menor ticket promedio, aproximadamente **$361.86 mil**.
- Las propiedades **comerciales** registran **1,071 ventas**, pero presentan el mayor ticket promedio, aproximadamente **$1.80 millones**.
- El análisis de cohortes muestra que existen **recompras posteriores a la primera adquisición**. La tasa mensual de recompra es variable y tiende a disminuir conforme aumenta el tiempo transcurrido desde la adquisición.
- El crecimiento de **11.14 % en 2024** evidencia una evolución interanual positiva, aunque los ingresos presentan fluctuaciones a lo largo del año.

### 🎯 Recomendaciones estratégicas

- **Mantener las casas como una categoría prioritaria**, debido a que representan el mayor ingreso acumulado del periodo.
- **Fortalecer el canal Corredor**, dado que concentra la mayor generación de ingresos, sin dejar de evaluar oportunidades para incrementar el desempeño del canal Directo.
- Desarrollar estrategias diferenciadas por tipo de propiedad, considerando el elevado volumen de ventas de los **departamentos** y el alto valor por operación de las propiedades **comerciales**.
- Implementar estrategias de **fidelización y recompra**, especialmente durante los primeros meses posteriores a la adquisición.
- Analizar las características del segmento **Primera vez**, debido a su elevado volumen de operaciones, para desarrollar estrategias orientadas a convertir estos compradores en clientes recurrentes.
